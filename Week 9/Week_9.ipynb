```python
print(df[['bedrooms', 'bathrooms', 'sqft_living', 'grade', 'yr_built']].head())
```

```python
df = df.drop_duplicates().dropna().copy()
```

```python
df = df[(df['bedrooms'] > 0) & (df['bathrooms'] > 0)].copy()
```

```python
df['date'] = pd.to_datetime(df['date'])
df['Sale_Year'] = df['date'].dt.year
df['Property_Age'] = df['Sale_Year'] - df['yr_built']

print(df[['date', 'Sale_Year', 'Property_Age']].head())
```

```python
df['date'] = pd.to_datetime(df['date'])
df['Sale_Year'] = df['date'].dt.year
df['Property_Age'] = df['Sale_Year'] - df['yr_built']
df['Is_Renovated'] = (df['yr_renovated'] > 0).astype(int)

print(df[['Sale_Year', 'Property_Age', 'Is_Renovated']].head())
```

```python
df['Price_Per_Sqft_Living'] = df['price'] / df['sqft_living']
df['Price_Per_Sqft_Lot'] = df['price'] / df['sqft_lot']

print(df[['price', 'sqft_living', 'sqft_lot',
          'Price_Per_Sqft_Living', 'Price_Per_Sqft_Lot']].head())
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['price'])
plt.title('House Price Outliers')
plt.show()
```

```python
Q1 = df['price'].quantile(0.25)
Q3 = df['price'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = df[(df['price'] < lower_bound) |
               (df['price'] > upper_bound)]

print("Number of Outliers:", len(outliers))
print(outliers[['price']].head())
```
